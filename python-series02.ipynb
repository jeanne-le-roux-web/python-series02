## Better version : second try

```python
import numpy as np
import matplotlib.pyplot as plt

def sinusoid(t, amplitude=1.0, frequency=2.0, phase=0.0):

    x = amplitude * np.sin(2 * np.pi * frequency * t + phase)
    return x
   

# 1. Definition temporel axis 
t = np.linspace(0, 1, 1000)

# 2. same amplitude but different phases (pi/2)
s1 = sinusoid(t, amplitude=1.0, frequency=1.0, phase=0.0)             # first signal 
s2 = sinusoid(t, amplitude=1.0, frequency=1.0, phase=np.pi / 2)       # Second signal 

# 3. display both signals to observe them 
plt.figure(figsize=(10, 5))
plt.plot(t, s1, label='Signal 1 (phase = 0)', color='blue')
plt.plot(t, s2, label='Signal 2 (phase = $\pi/2$)', color='orange')

# 4. Grid modifications and labels
plt.title("Sinusoidal function")
plt.xlabel("Temps t (s)")
plt.ylabel("Amplitude")
plt.axhline(0, color='grey', linewidth=0.8, linestyle=':') # Ligne d'équilibre à y=0
plt.legend(loc='upper right')
plt.grid(True)

# 5. Save the plot as a PDF file 
plt.savefig("sinusoid_phase_plot.pdf", format="pdf", bbox_inches='tight')
plt.show()
```

## 1.3. Final version : Create a notebook to plot sinusoidal signals 
In this final section, we refactor our code to follow high standards of software quality and research reproducibility. We centralize the mathematical computation inside a dedicated, well-documented function and generate comparative signals.

### Objectives Achieved:
- **Code modularity:** Creation of a reusable `sinusoid` function that accepts parameters and returns a NumPy array named `x`.
- **Signal generation:** Creation of two distinct arrays (`s1` and `s2`) utilizing different frequencies.
- **Code quality & readability:** Meaningful variable names, inline comments focusing on the *why*, and comprehensive docstrings.

```python
import numpy as np
    # numpy is for numerical calculations and is imported as np to use its functions and methods easily and faster
import matplotlib.pyplot as plt
    # matplotlib is for graphing and is imported as plt to use its functions and methods easily and faster


def sinusoid(t, amplitude=1.0, frequency=1.0, phase=0.0):
    
    # Mathematical modeling of the sine wave :
    x = amplitude * np.sin(2 * np.pi * frequency * t + phase)
    return x
    # return is used to integrate the function in the main code and to use it to generate the signals

# 1. Define a shared time vector from 0 to 1 second with 1000 smooth points (for high-resolution)
t = np.linspace(0, 1, 1000)
# linspace is using to generate number in a regular interval

# 2. Generate two separate sinusoidal signals (s1 and s2) with different frequencies using our function and different phases
s1 = sinusoid(t, amplitude=1.0, frequency=1.0)              # Sinus 1 (frequency = 1.0 Hz)
s2 = sinusoid(t, amplitude=1.0, frequency=2.0, phase=0.5)   # Sinus 2 (frequency = 2.0 Hz with phase shift)
    # s1 and s2 are the two signals generated by the sinusoid function with different frequencies and phases
    # t is the time vector used to generate the signals
    # amplitude is the maximum value of the signal
    # frequency is the number of cycles per second of the signal
    # phase is the initial angle of the signal in radians

# 3. Plot both signals as a function of time (t) and distinguish them with labels and colors 
plt.figure(figsize=(6, 3))
    # depend how many grid lines i want to show
plt.plot(t, s1, label='Sinus 1', color='tab:blue')
plt.plot(t, s2, label='Sinus 2', color='tab:orange')
    # plt to represent graphically the two signals and to compare them visually
    # t, s1, s2 are the variables to plot and label is used to add a legend to the plot 
    # t is the time vector
    # s1 and s2 are the two signals generated by the sinusoid function
    # tab:blue and tab:orange are the colors used to distinguish the two signals in the plot 
    # we used the tab: prefix to use the default color cycle of matplotlib

# 4. Enhance plot layout for academic clarity and visual accuracy
plt.title("Fonction sinusoïdale")
plt.xlabel("t")
    # "t" is used to represent time in seconds and to indicate the x-axis of the plot
plt.ylabel("Amplitude")
    # "Amplitude" is used to represent the y-axis of the plot and to indicate the amplitude of the signals
plt.xlim(0.0, 1)
    # "xlim" is udes to set the limits
plt.ylim(-1.05, 1.05)
plt.grid(True, linestyle='-', alpha=0.7)
    # "grid" is used to add a grid to the plot and to improve readability
    # true is used to show the grid 
    # linestyle is used to set the style of the grid lines
    # alpha is used to set the transparency of the grid lines between 0 and 1 (0 is fully transparent, 1 is fully opaque)
plt.legend(loc='upper right')
    # "legend" is used to add a legend to the plot and to indicate which signal is which
    # loc is used to set the location of the legend in the plot (upper right, upper left, lower right, lower left, center)

# 5. Export the final plot as a vector PDF file for publication quality
plt.savefig("sinusoid_final_plot.pdf", format="pdf", bbox_inches='tight')
    # savefig is used to save the plot as a PDF file and to export the plot as a vector file for publication quality
    # bbox_inches is used to set the bounding box of the plot and to remove any extra whitespace around the plot
plt.show()
    # show is used to display the plot in a window and to visualize the plot in the notebook
```
