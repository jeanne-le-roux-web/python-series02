## First try : 

```python
# 1. import libraries
# numpy is for numerical calculations, matplotlib is for graphing and as plt a surname for matplotlib.pyplot
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# 2. Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
# linspace is using to generate number in a regular interval
t = np.linspace(0, 1, 1000)
```

```python
# 3. Create a numpy array called s that contains the values of a sine function of t (s is for signal)
# ωt (avec ω=2πf)
s = np.sin(2 * np.pi * t)
```

```python
# 4. Plot s as a function of t  
plt.plot(t, s)
# 5. Add a title to the plot that says "Sinusoidal function"
plt.title("Sinusoidal curve")
# 6. Add a label to the x axis
# label to add names for axis 
plt.xlabel("Time (s)")
# 7. Add a label to the y axis
plt.ylabel("Amplitude")
# 8. Add a grid to the plot
# grid to add a grid to the plot and sometime 'true' is used to show the grid
plt.grid()
# 9. Save the plot as a png file called "sine_function.png"
# to save directly the plot as a png file in my folder 
plt.savefig("sine_function.png")
# sometimes plt.show() is used to show the plot in a classic python script 
```

## Better version : second try

```python
import numpy as np
import matplotlib.pyplot as plt

def sinusoid(t, amplitude=1.0, frequency=2.0, phase=0.0):

    s = amplitude * np.sin(2 * np.pi * frequency * t + phase)
    return s

# 1. Definition temporel axis 
t = np.linspace(0, 1, 1000)

# 2. Utilisation de notre fonction avec la même amplitude, mais une phase différente (décalage de pi/2)
s1 = sinusoid(t, amplitude=1.0, frequency=1.0, phase=0.0)             # Premier signal (sinus de base)
s2 = sinusoid(t, amplitude=1.0, frequency=1.0, phase=np.pi / 2)       # Second signal (déphasé de pi/2)

# 3. Tracé des deux signaux sur le même graphique pour observer le décalage
plt.figure(figsize=(10, 5))
plt.plot(t, s1, label='Signal 1 (phase = 0)', color='blue')
plt.plot(t, s2, label='Signal 2 (phase = $\pi/2$)', color='orange')

# 4. Habillage du graphique pour assurer une lisibilité optimale
plt.title("Sinusoidal function")
plt.xlabel("Temps t (s)")
plt.ylabel("Amplitude")
plt.axhline(0, color='grey', linewidth=0.8, linestyle=':') # Ligne d'équilibre à y=0
plt.legend(loc='upper right')
plt.grid(True)

# 5. Sauvegarde de la version finale propre au format PDF
plt.savefig("sinusoid_phase_plot.pdf", format="pdf", bbox_inches='tight')
plt.show()
```
