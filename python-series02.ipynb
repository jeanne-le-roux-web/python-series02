The goal is to code the generation of a sine wave and its visualization, and to pay attention to how to code, describe and reference. 

The formule of a sine wave : 
y(t)=A⋅sin⁡(ωt+φ)


```python
# 1. import libraries
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# 2. Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0, 1, 1000)
```

```python
# 3. Create a numpy array called s that contains the values of a sine function of t (s is for signal)
s = np.sin(2 * np.pi * t)
```

```python
# 4. Plot s as a function of t  
plt.plot(t, s)
# 5. Add a title to the plot that says "Sinusoidal function"
plt.title("Sinusoidal function")
```
