The goal is to code the generation of a sine wave and its visualization, and to pay attention to how to code, describe and reference. 

The formule of a sine wave : 
y(t)=A⋅sin⁡(ωt+φ)


```python
# 1. import libraries
# numpy is for numerical calculations, matplotlib is for graphing and as plt a surname for matplotlib.pyplot
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# 2. Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
# linspace is using to generate number in a regular interval
t = np.linspace(0, 1, 1000)
```

```python
# 3. Create a numpy array called s that contains the values of a sine function of t (s is for signal)
# ωt (avec ω=2πf)
s = np.sin(2 * np.pi * t)
```

```python
# 4. Plot s as a function of t  
plt.plot(t, s)
# 5. Add a title to the plot that says "Sinusoidal function"
plt.title("Sinusoidal function")
# 6. Add a label to the x axis
# label to add names for axis 
plt.xlabel("Time (s)")
# 7. Add a label to the y axis
plt.ylabel("Amplitude")
# 8. Add a grid to the plot
# grid to add a grid to the plot and sometime 'true' is used to show the grid
plt.grid()
# 9. Save the plot as a png file called "sine_function.png"
# to save directly the plot as a png file in my folder 
plt.savefig("sine_function.png")
# sometimes plt.show() is used to show the plot in a classic python script 
```
