## Better version : second try

```python
import numpy as np
import matplotlib.pyplot as plt

def sinusoid(t, amplitude=1.0, frequency=2.0, phase=0.0):

    x = amplitude * np.sin(2 * np.pi * frequency * t + phase)
    return x
   

# 1. Definition temporel axis 
t = np.linspace(0, 1, 1000)

# 2. same amplitude but different phases (pi/2)
s1 = sinusoid(t, amplitude=1.0, frequency=1.0, phase=0.0)             # first signal 
s2 = sinusoid(t, amplitude=1.0, frequency=1.0, phase=np.pi / 2)       # Second signal 

# 3. display both signals to observe them 
plt.figure(figsize=(10, 5))
plt.plot(t, s1, label='Signal 1 (phase = 0)', color='blue')
plt.plot(t, s2, label='Signal 2 (phase = $\pi/2$)', color='orange')

# 4. Grid modifications and labels
plt.title("Sinusoidal function")
plt.xlabel("Temps t (s)")
plt.ylabel("Amplitude")
plt.axhline(0, color='grey', linewidth=0.8, linestyle=':') # Ligne d'équilibre à y=0
plt.legend(loc='upper right')
plt.grid(True)

# 5. Save the plot as a PDF file 
plt.savefig("sinusoid_phase_plot.pdf", format="pdf", bbox_inches='tight')
plt.show()
```

## 1.3. Final version : Create a notebook to plot sinusoidal signals 
In this final section, we refactor our code to follow high standards of software quality and research reproducibility. We centralize the mathematical computation inside a dedicated, well-documented function and generate comparative signals.

### Objectives Achieved:
- **Code modularity:** Creation of a reusable `sinusoid` function that accepts parameters and returns a NumPy array named `x`.
- **Signal generation:** Creation of two distinct arrays (`s1` and `s2`) utilizing different frequencies.
- **Code quality & readability:** Meaningful variable names, inline comments focusing on the *why*, and comprehensive docstrings.

```python
import numpy as np
import matplotlib.pyplot as plt

def sinusoid(t, amplitude=1.0, frequency=1.0, phase=0.0):
    """
    Computes and returns a sinusoidal signal for a given time vector.
    
    Why: Centralizing the mathematical formula inside a single function prevents code 
    duplication, ensures consistency, and allows easy parameter adjustments 
    (amplitude, frequency, phase) without modifying the plotting logic.
    
    Parameters:
    - t (numpy.ndarray): Time vector.
    - amplitude (float): Maximum peak amplitude of the wave.
    - frequency (float): Frequency of the signal in Hertz (Hz).
    - phase (float): Initial phase shift in radians.
    
    Returns:
    - x (numpy.ndarray): The computed sinusoidal signal values.
    """
    # Mathematical modeling of the sine wave; the output array is explicitly named 'x' as requested
    x = amplitude * np.sin(2 * np.pi * frequency * t + phase)
    return x

# 1. Define a shared time vector from 0 to 1 second with 1000 smooth points
t = np.linspace(0, 1, 1000)

# 2. Generate two separate NumPy arrays (s1 and s2) with different frequencies using our function
s1 = sinusoid(t, amplitude=1.0, frequency=1.0)              # Sinus 1 (frequency = 1.0 Hz)
s2 = sinusoid(t, amplitude=1.0, frequency=2.0, phase=0.5)   # Sinus 2 (frequency = 2.0 Hz with phase shift)

# 3. Plot both signals as a function of time (t)
plt.figure(figsize=(8, 5))
plt.plot(t, s1, label='Sinus 1', color='tab:blue')
plt.plot(t, s2, label='Sinus 2', color='tab:orange')

# 4. Enhance plot layout for academic clarity and visual accuracy
plt.title("Fonction sinusoïdale")
plt.xlabel("t")
plt.ylabel("Amplitude")
plt.xlim(0.0, 1.0)
plt.ylim(-1.05, 1.05)
plt.grid(True, linestyle='-', alpha=0.6)
plt.legend(loc='upper right')

# 5. Export the final plot as a vector PDF file for publication quality
plt.savefig("sinusoid_final_plot.pdf", format="pdf", bbox_inches='tight')
plt.show()
```
